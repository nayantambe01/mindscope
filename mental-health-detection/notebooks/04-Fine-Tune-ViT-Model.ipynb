```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from transformers import TFViTForImageClassification
import numpy as np
import albumentations as A

# --- 1. Define Paths and Parameters ---
base_dir = '../data/FER2013_images/train'
CLASSES_TO_USE = ['sad', 'neutral']
# Vision Transformers expect a larger input size and 3 color channels (RGB)
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 16 # Use a smaller batch size for large models

# --- 2. Create an Advanced Data Augmentation Pipeline ---
# This creates more training data by slightly altering the images (rotating, flipping)
# which makes the final model more robust.
transform = A.Compose([
    A.HorizontalFlip(p=0.5),
    A.Rotate(limit=10, p=0.5),
    A.RandomBrightnessContrast(p=0.2),
])

def augment_data(image):
    # The ViT model expects 3 color channels, so we convert grayscale to RGB
    image_rgb = tf.image.grayscale_to_rgb(tf.convert_to_tensor(image))
    augmented = transform(image=image_rgb.numpy())
    return augmented['image']

# Use ImageDataGenerator to load images and apply our augmentation
datagen = ImageDataGenerator(
    rescale=1./255,
    preprocessing_function=augment_data,
    validation_split=0.2
)

train_generator = datagen.flow_from_directory(
    base_dir,
    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    classes=CLASSES_TO_USE,
    subset='training'
)

validation_generator = datagen.flow_from_directory(
    base_dir,
    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    classes=CLASSES_TO_USE,
    subset='validation'
)

# --- 3. Build the Fine-Tuned ViT Model ---
# Load a powerful, pre-trained Vision Transformer from Hugging Face
MODEL_NAME = 'google/vit-base-patch16-224'

# We load the model, telling it we have 2 labels and giving our labels names.
model = TFViTForImageClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2,
    id2label={0: "neutral", 1: "sad"},
    label2id={"neutral": 0, "sad": 1}
)

# --- 4. Compile and Train the Model ---
# We use a lower learning rate for fine-tuning, which is a best practice.
optimizer = tf.keras.optimizers.Adam(learning_rate=5e-5)
loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
metric = tf.keras.metrics.SparseCategoricalAccuracy('accuracy')

model.compile(optimizer=optimizer, loss=loss, metrics=[metric])

print("\nStarting to fine-tune the Vision Transformer model...")
# The .fit() function starts the training process.
history = model.fit(
    train_generator,
    epochs=3, # Fine-tuning requires fewer epochs
    validation_data=validation_generator
)
print("Model fine-tuning complete.")

# --- 5. Save the New, State-of-the-Art Model ---
# We save it in the standard Hugging Face format.
model.save_pretrained('../models/vit-emotion-model-finetuned')
print("\n✅ New, state-of-the-art ViT model saved to models/vit-emotion-model-finetuned")

```
